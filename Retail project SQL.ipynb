```python
import pandas as pd
import sqlite3

df = pd.read_excel('cleaned_supermarket_sales (1).xlsx')

conn = sqlite3.connect('supermarket.db')

df.to_sql('sales', conn, if_exists='replace', index=False)

print("SQL Database Created Successfully!")
```

```python
query = "SELECT * FROM sales LIMIT 10"

result = pd.read_sql_query(query, conn)

print(result)
```

```python
query = """
SELECT *
FROM sales
WHERE Total > 1000
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT *
FROM sales
ORDER BY Total DESC
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT City, SUM(Total) AS Total_Sales
FROM sales
GROUP BY City
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT
SUM(Total) AS Total_Sales,
AVG(Total) AS Average_Sales
FROM sales
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
CREATE TABLE IF NOT EXISTS sales_summary (
    City TEXT,
    Total_Sales REAL
)
"""

conn.execute(query)
conn.commit()

print("Table Created Successfully!")
```

```python
query = """
SELECT sales.City, sales.Total, sales_summary.Total_Sales
FROM sales
LEFT JOIN sales_summary
ON sales.City = sales_summary.City
LIMIT 10
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT sales.City, sales.Total,
sales_summary.Total_Sales
FROM sales
LEFT JOIN sales_summary
ON sales.City = sales_summary.City
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
SELECT *
FROM sales
WHERE Total > (
    SELECT AVG(Total)
    FROM sales
)
"""

result = pd.read_sql_query(query, conn)
print(result)
```

```python
query = """
CREATE VIEW IF NOT EXISTS city_sales AS
SELECT City, SUM(Total) AS Total_Sales
FROM sales
GROUP BY City
"""

conn.execute(query)
conn.commit()

print("View Created Successfully!")
```

```python
query = "SELECT * FROM city_sales"

result = pd.read_sql_query(query, conn)
print(result)
```

```python
conn.close()

print("SQL Analysis Completed Successfully!")
```
