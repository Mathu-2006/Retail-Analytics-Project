```python
import pandas as pd

df = pd.read_excel('cleaned_supermarket_sales (1).xlsx')

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns)
print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.describe()
```

```python
df.groupby('Product line')['Total'].sum().sort_values(ascending=False)
```

```python
df.groupby('City')['Total'].sum()
```

```python
df.groupby('Gender')['Total'].sum()
```

```python
df.groupby('Payment')['Total'].sum()
```

```python
import matplotlib.pyplot as plt

df.groupby('Product line')['Total'].sum().plot(
    kind='bar', figsize=(10,5), title='Product Line Wise Sales'
)

plt.xlabel('Product Line')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.groupby('City')['Total'].sum().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6),
    title='City Wise Sales Analysis'
)

plt.ylabel('')
plt.show()
```

```python
df.groupby('Gender')['Total'].sum().plot(
    kind='bar',
    color=['skyblue', 'pink'],
    title='Gender Wise Sales Analysis',
    figsize=(6,4)
)

plt.xlabel('Gender')
plt.ylabel('Total Sales')
plt.show()
```

```python
df.groupby('Payment')['Total'].sum().plot(
    kind='bar',
    title='Payment Method Wise Sales',
    figsize=(6,4)
)

plt.xlabel('Payment Method')
plt.ylabel('Total Sales')
plt.show()
```
